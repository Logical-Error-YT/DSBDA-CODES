```python
# Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
df = sns.load_dataset('iris')
```

```python

# Display first 5 rows
print(df.head())
```

```python
# 1. Features and their types

print("\nFeatures and Data Types:\n")
print(df.dtypes)
```

```python
# 2. Histogram for each feature

df.hist(figsize=(10,8))

plt.suptitle("Histogram of Iris Dataset Features")
plt.show()

```

```python
# 3. Box plot for each feature

plt.figure(figsize=(12,6))

# Box plot for all numeric columns
sns.boxplot(data=df)

plt.title("Box Plot of Iris Dataset Features")
plt.show()

```

```python
# 4. Statistical Summary

print("\nStatistical Summary:\n")
print(df.describe())
```
