```python
import pandas as pd

# Create sample dataset
data = {
    'Age_Group': ['Young', 'Young', 'Adult', 'Adult', 'Senior', 'Senior'],
    'Income': [20000, 25000, 40000, 42000, 30000, 28000]
}

df = pd.DataFrame(data)
```

```python
# Group by Age_Group and calculate statistics
result = df.groupby('Age_Group')['Income'].agg(
    ['mean', 'median', 'min', 'max', 'std']
)
print("Summary Statistics:\n", result)
```

```python
# Create list of values for each group
grouped_list = df.groupby('Age_Group')['Income'].apply(list)

print("\nIncome values for each Age Group:\n", grouped_list)
```
