# Descriptive Statistics — Measures of Central Tendency and Variability

This notebook covers:
1. Summary statistics of numeric variables grouped by a categorical variable
2. Basic statistical details for the Iris dataset species

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

print('Libraries imported!')
```

## Part 1: Summary Statistics Grouped by Categorical Variable

We create a sample dataset with `AgeGroup` (categorical) and `Income` (numeric).

`groupby()` groups the data by a categorical column, and `.agg()` applies multiple aggregation functions at once.

```python
np.random.seed(0)
n = 200
age_groups = np.random.choice(['18-25', '26-35', '36-45', '46-60'], n)
income = np.where(age_groups == '18-25', np.random.randint(20000, 40000, n),
         np.where(age_groups == '26-35', np.random.randint(35000, 70000, n),
         np.where(age_groups == '36-45', np.random.randint(55000, 90000, n),
                                          np.random.randint(40000, 80000, n))))

df = pd.DataFrame({'AgeGroup': age_groups, 'Income': income})
df.head()
```

```python
# Group by AgeGroup and compute summary stats for Income
summary = df.groupby('AgeGroup')['Income'].agg(
    Mean='mean',
    Median='median',
    Std='std',
    Min='min',
    Max='max'
).round(2)

print('=== Income Summary Statistics by Age Group ===')
summary
```

```python
# List of numeric values per age group
group_list = {grp: list(data['Income']) for grp, data in df.groupby('AgeGroup')}
for grp, vals in group_list.items():
    print(f'{grp}: First 5 values -> {vals[:5]}')
```

## Part 2: Statistical Details of Iris Dataset

The Iris dataset has 3 species: Iris-setosa, Iris-versicolor, Iris-virginica.

We compute percentile, mean, std for each species using `describe()`.

```python
# Load Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
iris_df.head()
```

```python
# Statistical details per species
for species in iris_df['species'].unique():
    print(f'\n=== {species} ===')
    print(iris_df[iris_df['species'] == species].describe().round(3))
```
