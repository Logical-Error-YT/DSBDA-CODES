# Text Analytics

This notebook demonstrates core NLP preprocessing steps:
1. **Tokenization** — Split text into words/sentences
2. **POS Tagging** — Identify grammatical role of each word
3. **Stop Words Removal** — Remove common uninformative words
4. **Stemming** — Reduce word to its root form (e.g., "running" → "run")
5. **Lemmatization** — Reduce word to its dictionary form
6. **TF-IDF** — Term Frequency × Inverse Document Frequency representation

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

print('Libraries ready!')
```

## Sample Document

```python
text = """Natural Language Processing is a fascinating field of Artificial Intelligence.
It enables computers to understand and interpret human language.
Many applications like chatbots and translators are built using NLP techniques."""

print('Document:')
print(text)
```

## Step 1: Tokenization

`word_tokenize()` splits text into individual words. `sent_tokenize()` splits text into sentences.

```python
word_tokens = word_tokenize(text)
sent_tokens = sent_tokenize(text)
print('Word Tokens:', word_tokens[:15], '...')
print('\nSentence Tokens:')
for s in sent_tokens:
    print(' -', s)
```

## Step 2: POS Tagging

Assigns grammatical tags: NN=Noun, VBZ=Verb, JJ=Adjective, IN=Preposition, etc.

```python
pos_tags = pos_tag(word_tokens)
print('POS Tags (first 15):')
print(pos_tags[:15])
```

## Step 3: Stop Words Removal

Stop words like 'is', 'the', 'and' carry little meaning. We remove them to focus on important words.

```python
stop_words = set(stopwords.words('english'))
filtered = [w for w in word_tokens if w.lower() not in stop_words and w.isalpha()]
print('After removing stop words:')
print(filtered)
```

## Step 4: Stemming

PorterStemmer reduces words to their base/stem (e.g., 'applications' → 'applic'). Fast but may create non-real words.

```python
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered]
print('Stemmed words:')
print(stemmed)
```

## Step 5: Lemmatization

WordNetLemmatizer returns proper dictionary forms (e.g., 'enables' → 'enable'). Better than stemming linguistically.

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered]
print('Lemmatized words:')
print(lemmatized)
```

## Step 6: TF-IDF Representation

**TF (Term Frequency):** How often a word appears in a document.
**IDF (Inverse Document Frequency):** Penalizes words that appear in many documents (common words).
**TF-IDF = TF × IDF** — gives high scores to words that are frequent in one doc but rare overall.

```python
docs = sent_tokens  # Use sentences as documents
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(docs)

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)
print('TF-IDF Matrix:')
tfidf_df.round(3)
```
