# Data Visualization I — Titanic Dataset

**Dataset:** Titanic (inbuilt in Seaborn)

**Tasks:**
1. Explore patterns using Seaborn
2. Plot a histogram of ticket fare distribution

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load Titanic dataset from Seaborn
df = sns.load_dataset('titanic')
print('Shape:', df.shape)
df.head()
```

## Explore Patterns: Survival by Class

A count plot shows how many passengers survived vs died across each passenger class (1st, 2nd, 3rd). We can clearly see 1st class had higher survival.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='pclass', hue='survived', palette='Set2')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(['Not Survived', 'Survived'])
plt.tight_layout()
plt.show()
```

## Explore Patterns: Survival by Gender

Females had a significantly higher survival rate than males, reflecting the 'women and children first' policy.

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='sex', hue='survived', palette='pastel')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(['Not Survived', 'Survived'])
plt.tight_layout()
plt.show()
```

## Part 2: Histogram of Fare Distribution

A histogram shows how the `fare` column is distributed. The x-axis represents fare ranges (bins), and the y-axis shows how many passengers fall in each range.

**Observation:** Most passengers paid low fares, with a few outliers paying very high fares — indicating a right-skewed distribution.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['fare'].dropna(), bins=30, color='steelblue', edgecolor='black')
plt.title('Distribution of Ticket Fare')
plt.xlabel('Fare (£)')
plt.ylabel('Number of Passengers')
plt.axvline(df['fare'].mean(), color='red', linestyle='--', label=f'Mean: {df["fare"].mean():.1f}')
plt.axvline(df['fare'].median(), color='green', linestyle='--', label=f'Median: {df["fare"].median():.1f}')
plt.legend()
plt.tight_layout()
plt.show()

print(f'Mean Fare  : {df["fare"].mean():.2f}')
print(f'Median Fare: {df["fare"].median():.2f}')
print(f'Std Dev    : {df["fare"].std():.2f}')
```
