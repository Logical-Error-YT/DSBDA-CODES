# Data Visualization II — Box Plot: Age vs Gender (Titanic)

**Dataset:** Titanic (inbuilt Seaborn)

**Task:** Plot a box plot of age distribution grouped by gender and survival status.

**Observations to derive:** Were children more likely to survive? Were women younger/older than men on average?

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = sns.load_dataset('titanic')
print('Dataset loaded. Shape:', df.shape)
df[['sex', 'age', 'survived']].describe()
```

## Box Plot: Age Distribution by Gender and Survival

A **box plot** shows:
- The **box** = Interquartile range (Q1 to Q3, middle 50% of data)
- The **line inside box** = Median
- **Whiskers** = Range beyond IQR (up to 1.5×IQR)
- **Dots** = Outliers

Here:
- `x='sex'` groups by gender
- `y='age'` plots age distribution
- `hue='survived'` splits each gender into survived (1) vs not survived (0)

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='sex', y='age', hue='survived', palette='Set3')
plt.title('Age Distribution by Gender and Survival Status')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()
```

## Observations

Based on the box plot:

```python
# Compute median ages for each group
obs = df.groupby(['sex', 'survived'])['age'].median().reset_index()
obs.columns = ['Gender', 'Survived', 'Median Age']
obs['Survived'] = obs['Survived'].map({0: 'No', 1: 'Yes'})
print('=== Median Age by Gender and Survival ===')
obs
```

## Written Observations

1. **Female survivors** tend to have a lower median age than female non-survivors, suggesting younger females had higher survival rates.

2. **Male non-survivors** have a slightly higher median age than male survivors — older men were less likely to survive.

3. **Female passengers** overall had a much higher survival rate (wider 'Yes' box for females) compared to males.

4. There are **outliers** in both genders with very high ages (70+), mostly non-survivors.

5. The interquartile ranges overlap considerably, indicating age alone doesn't fully predict survival — gender plays a larger role.

6. Male non-survivors show a wider age spread, indicating men of all ages were affected uniformly.
