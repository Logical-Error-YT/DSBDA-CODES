# Data Visualization III — Iris Dataset

**Dataset:** Iris (sklearn built-in)

**Source:** https://archive.ics.uci.edu/ml/datasets/Iris

**Tasks:**
1. List features and their types
2. Histogram for each feature
3. Box plot for each feature
4. Compare distributions and identify outliers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]
print('Shape:', df.shape)
df.head()
```

## Step 1: Features and Their Types

The Iris dataset has 4 numeric (continuous) features and 1 categorical feature.

```python
feature_info = pd.DataFrame({
    'Feature': df.columns,
    'Type': df.dtypes.values,
    'Variable Type': ['Numeric (continuous)'] * 4 + ['Categorical (nominal)']
})
print(feature_info.to_string(index=False))
```

## Step 2: Histograms for Each Feature

Histograms show the **frequency distribution** of each numeric feature. They help us understand whether data is normally distributed, skewed, or bimodal.

```python
numeric_cols = iris.feature_names
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
axes = axes.flatten()

colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52']
for i, col in enumerate(numeric_cols):
    axes[i].hist(df[col], bins=15, color=colors[i], edgecolor='black', alpha=0.8)
    axes[i].set_title(f'Histogram: {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

plt.suptitle('Histograms of Iris Features', fontsize=14)
plt.tight_layout()
plt.show()
```

## Step 3: Box Plots for Each Feature

Box plots show the **spread and outliers** in each feature. The box = IQR (Q1–Q3), the line = median, whiskers = 1.5×IQR, and dots = outliers.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    sns.boxplot(data=df, x='species', y=col, palette='Set2', ax=axes[i])
    axes[i].set_title(f'Box Plot: {col}')

plt.suptitle('Box Plots of Iris Features by Species', fontsize=14)
plt.tight_layout()
plt.show()
```

## Step 4: Compare Distributions and Identify Outliers

We use a pair plot to compare all features simultaneously and identify patterns and outliers.

```python
sns.pairplot(df, hue='species', palette='Set1', diag_kind='kde')
plt.suptitle('Pair Plot — Iris Dataset', y=1.02)
plt.show()
```

```python
# Count outliers using IQR method for each feature
print('=== Outlier Count per Feature (IQR method) ===')
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5*IQR) | (df[col] > Q3 + 1.5*IQR)]
    print(f'{col}: {len(outliers)} outliers')
```

## Inferences

1. **Setosa** is clearly separable from other species in petal length and petal width.
2. **Versicolor and Virginica** overlap more — harder to separate.
3. **Sepal width** has the most outliers, especially in setosa.
4. **Petal length and width** show the most species separation — best features for classification.
5. Distributions are approximately normal for most features within each species.
